```python
# Load and display the single image
img = cv2.imread(MULTI_SHERD_SINGLE_IMAGE)
assert img is not None, f"Could not load image: {MULTI_SHERD_SINGLE_IMAGE}"

plt.figure(figsize=(12, 10))
imshow_cm(plt.gca(), img[:, :, ::-1])
plt.title(f"Original Scan  ({img.shape[1]}x{img.shape[0]} px)")
plt.show()
```
